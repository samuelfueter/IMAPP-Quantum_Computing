```python
from qiskit import QuantumCircuit
from qiskit.synthesis.qft import synth_qft_full
from qiskit import quantum_info as qi
from qiskit.visualization import array_to_latex
from qiskit_aer import Aer
```

```python
# Let's create a quantum circuit with 2 qubits
qc = QuantumCircuit(2)
# We initialize the qubits to the state |00> and then apply the QFT circuit to them
qc.initialize('11', qc.qubits)
# Apply the QFT circuit to the qubits
qc.compose(synth_qft_full(2), qubits = qc.qubits, inplace = True)
```

```python
qc.draw()
```

```python
# The professor wrote: psi_f = qi.Statevector.from_circuit(qc), but mine doesn't work.
psi_f = qi.Statevector.from_instruction(qc) 
array_to_latex(psi_f)
```

```python
# Computing all the possible states of the circuit, the following QFT of each state of 
# the circuit of 2 qubits are the following:
    # |00> -> 1/2(|00> + |01> + |10> + |11>)
    # |01> -> 1/2(|00> + i|01> - |10> - i|11>)
    # |10> -> 1/2(|00> - |01> + |10> - |11>)
    # |11> -> 1/2(|00> - i|01> - |10> + i|11>)
```

```python
backend = Aer.get_backend('statevector_simulator')
psi_f = backend.run(qc).result().get_statevector()
array_to_latex(psi_f)
```
